```python
#Manipulación de datos y Aplicación de funciones y expresiones
import functools
#APLICACION DEL MODULO 3 INICIA AQUÍ
#M3 Esta libreria sirve para crear y personalizar diagramas en python
import matplotlib.pyplot as plt
from matplotlib.patheffects import withStroke
import scipy.stats as stats
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix
from sklearn.metrics import confusion_matrix, roc_curve, auc
import seaborn as sns
#Encontre esta funcion para limpiar pantalla
from google.colab import output
def limpiar_pantalla():
    output.clear()
#Creación funcion de mi menu para mostrar informe
def mostrar_menu():
    print("\n**********************************************************************")
    print("      SISTEMA DE ANÁLISIS DE CIBERSEGURIDAD EN MÉXICO  ")
    print("**********************************************************************")
    print("1. Porcentaje de la población total referente a si toma o no medidas de seguridad")
    print("2. Medidas de ciberseguridad")
    print("3. Ciberdelitos")
    print("4. Regiones de México y su cultura de prevención (%) con Mapa Coroplético")
    print("5. Cultura de seguridad por sexo (%)")
    print("6. Efectividad de las medidas y análisis de distribución (boxplot)")
    print("7. Percepción vs Realidad ")
    print("8. Salir del programa")
    print("*"*70)
#Usare la funcion Lambda para dar formato a los números de población, como seran resultados de millones si los dejo de tipo entero sera dificil de leer con claridad
#y si los convierto a float me dara el numero con comas y sera mas sencillo de leer a simple vista
poblacion_formato= lambda x: f"{x:,.0f} personas"
ejecutando=True
while ejecutando:
    limpiar_pantalla()
    mostrar_menu()
    seleccion = input("Selecciona una opción del reporte (1-8): ")

    # Use un try directamente al MATCH-CASE para capturar errores al instante y asi tener un buen procesamiento en mis datos
    try:
        opcion = int(seleccion)

        match opcion:
            case 1:
                #Porcentaje de la poblacion total que si toman medidas de seguridad
                poblacion_si = df_ciberseguridad[df_ciberseguridad['realiza_medidas_seguridad'] == 1]['factor_expansion'].sum()
                poblacion_total = df_ciberseguridad['factor_expansion'].sum()
                #M3 Estadística descriptiva
                poblacion_no=poblacion_total-poblacion_si
                porcentaje = (poblacion_si / poblacion_total) * 100
                porcentaje_no = (poblacion_no / poblacion_total) * 100
                limpiar_pantalla()
                print("="*100)
                print("\n[OPCIÓN 1]PORCENTAJE DE LA POBLACIÓN TOTAL REFERENTE A SI TOMA O NO MEDIDAS DE SEGURIDAD:\n")
                print("="*100)
                #M3 Reporte de Estadística Descriptiva
                print(f"\nPoblación total nacional estimada: {poblacion_formato(poblacion_total)}")
                print(f"Población estimada que SÍ toma medidas: {poblacion_formato(poblacion_si)} ({porcentaje:.2f}%)")
                print(f"Población estimada que NO toma medidas: {poblacion_formato(poblacion_no)} ({porcentaje_no:.2f}%)")
                #M3 Visualización de Datos
                #Generare una grafica de pastel, porque es para mis datos de si y no es mas visible y entendible este tipo de grafico.
                plt.figure(figsize=(8, 8))
                plt.pie([poblacion_si, poblacion_no], labels=['Sí toman medidas', 'No toman medidas'], autopct='%1.1f%%', startangle=90, colors=['#88E788', '#FF9999'], explode=(0.05, 0))
                plt.title('Porcentaje de la Población que toma medidas de seguridad para proteger su computadora, tablet, celular o cuentas de internet en México', fontsize=12, fontweight='bold')
                plt.tight_layout()
                plt.show()
                plt.pause(0.1) # <--- Esta linea le da tiempo al sistema de procesar el gráfico largo (le tuve que pedir ayuda a la IA porque no salia mi input por el tamaño del grafico)
                input("\nPresione Enter para continuar y regresar al menú...")

            case 2:
                limpiar_pantalla()
                print("="*100)
                print("\n[OPCIÓN 2]MEDIDAS DE CIBERSEGURIDAD:\n")
                print("="*100)
              #Con la funcion filter nos ayuda a filtrar todas las columnas que contienen la palabra medida
                columnas_medidas = df_ciberseguridad.filter(like='medida_').columns
              #Creo un diccionario vacio para que se guarde mi informacion de cada columna por la que pasara mi ciclo y asi obtengo el total de cada columna con el si(1)
                totales_calculados = {}
                for col in columnas_medidas:
                   totales_calculados[col] = df_ciberseguridad[df_ciberseguridad[col] == 1]['factor_expansion'].sum()
              #Despues creo una serie para que me guarde mis valores de mi diccionario y los ordeno de  mayor a menor
                conteo_medidas = pd.Series(totales_calculados).sort_values(ascending=False)
              #Creo un diccionario para saber mas visualmante el nombre de la medida
                nombres_medidas = {
                        'medida_crear_contrasenas': 'Crear contraseñas seguras',
                        'medida_instalar_antivirus': 'Instalar o actualizar antivirus',
                        'medida_bloquear_ventanasemer': 'Bloquear ventanas emergentes',
                        'medida_cambiar_contrasenas': 'Cambiar contraseñas periódicamente',
                        'medida_evitar_sitios_inseguros': 'Evitar páginas web inseguras',
                        'medida_no_abrir_archivos_descono': 'No abrir archivos de remitentes desconocidos',
                        'medida_no_publicar_datos_redes': 'Evitar publicar datos personales en redes sociales',
                        'medida_otra_seguridad':'Usa otra medida que no fue mencionada anteriormente'
                    }
                #Transformación con map() (El diccionario traductor), ahora yo quiero ponerles un nombre mas legible a mis columnas de mi serie
                #Así que usaremos map()
                conteo_medidas.index = conteo_medidas.index.map(nombres_medidas)
                # Ahora ya que tenemos nuestra serie con sus nombres correspondientes podemos buscar la medida más utilizada (idxmax me ayuda a traer el dato mas alto de mi serie)
                medida_frecuente = conteo_medidas.idxmax()
                total_medida_victimas = conteo_medidas[medida_frecuente]
                #M3 Visualización de datos
                #Tabla de Frecuencias
                #Tabla de distribución de frecuencias para saber el total de veces que la población aplica cada medida
                #conteo_medidas ya viene siendo mi tabla de frecuencias, porque tengo los nombres y la cantidad de cada medida utilizada
                print("\nTABLA DE FRECUENCIA\n")
                print(conteo_medidas)
                #Ahora una grafica de barras para mejor visualización de los datos.
                conteo_medidas.plot(kind="barh", figsize=(8,4), color='#F1D3D6',fontsize=10)
                #plt.title= Añadir un titulo principal
                plt.title("Cantidad total de la población que usa medidas de seguridad en México\npara proteger su computadora, tablet, celular o cuentas de internet.", fontsize=12, pad=15)
                #Eje Y Intervalos o los bins
                #plt.ylabel Titulo a eje Y
                plt.ylabel("Medidas de seguridad",color='#00008B',fontsize=12)
                #Eje X  Frecuencia
                #plt.xlabel Titulo a eje X
                plt.xlabel("Cantidad",color='#00008B',fontsize=12)
                # Desactiva la notación científica (el 1e7)
                plt.ticklabel_format(style='plain', axis='x',useOffset=False)
                # Agrega comas de separación a los números del eje X para que sean fáciles de leer
                plt.gca().xaxis.set_major_formatter(plt.FuncFormatter(lambda x, pos: f"{int(x):,}"))
                #plt.show Mostrar el grafico en la pantalla
                plt.show()
                plt.pause(0.1)
                input("\nPresione Enter para continuar y regresar al menú...")
            case 3:
                #Ciberdelitos
                limpiar_pantalla()
                print("="*100)
                print("\n[OPCIÓN 3]CIBERDELITOS :\n")
                print("="*100)
                #Con la funcion filter nos ayuda a filtrar todas las columnas que contienen la palabra incidente_
                columnas_incidentes = df_ciberseguridad.filter(like='incidente_').columns
                #Creo un diccionario vacio para que se guarde mi informacion de cada columna por la que pasara mi ciclo y asi obtengo el total de cada columna con el si(1)
                totales_calculados = {}
                for col in columnas_incidentes:
                   totales_calculados[col] = df_ciberseguridad[df_ciberseguridad[col] == 1]['factor_expansion'].sum()
                #Ahora creo mi serie ya que tengo mis valores totales
                conteo_incidentes= pd.Series(totales_calculados).sort_values(ascending=False)
                #diccionario de nombres de los ciberdelitos
                nombres_ciberdelitos = {
                      'incidente_hackeo_correo': 'Hackeo de correo electrónico',
                      'incidente_hackeo_redes': 'Hackeo de redes sociales',
                      'incidente_ransomware': 'Secuestro de datos (Ransomware)',
                      'incidente_malware_conperdida': 'Malware con pérdida financiera o de datos',
                      'incidente_malware_general': 'Malware o virus general'
                  }
                #Transformación con map() (El diccionario traductor), ahora yo quiero ponerles un nombre mas legible a mis columnas de mi serie
                #Así que usaremos map()
                conteo_incidentes.index = conteo_incidentes.index.map(nombres_ciberdelitos)
                #Con idxmax buscamos de golpe cuál llave tiene el número de personas más alto
                ciberdelito_frecuente= conteo_incidentes.idxmax()
                total_victimas = conteo_incidentes[ciberdelito_frecuente]
                #M3 Visualización de datos
                #Se crea un grafico de barras , para mejor vizualizacion de los datos.
                conteo_incidentes.plot(kind="bar", figsize=(8,4),color='#F88379')
                #plt.title= Añadir un titulo principal
                plt.title("Ciberdelitos en México ",fontsize=12)
                #Eje X Intervalos o los bins
                #plt.xlabel Titulo a eje X
                plt.xlabel("Ciberdelito")
                #Eje Y  Frecuencia
                #plt.ylabel Titulo a eje Y
                plt.ylabel("Cantidad de afectados")
                # Desactiva la notación científica (el 1e7)
                plt.ticklabel_format(style='plain', axis='y')
                #Se agregan comas de separación a los números del eje Y para que sean fáciles de leer
                plt.gca().get_yaxis().set_major_formatter(plt.FuncFormatter(lambda x, loc: "{:,}".format(int(x))))
                #plt.show Mostrar el grafico en la pantalla
                plt.show()
                plt.pause(0.1)
                print("\nCONCLUSIÓN\n")
                print(f"\nEl ciberdelito más frecuente en el país es: {ciberdelito_frecuente}")
                print(f"\nTotal de víctimas estimadas: {poblacion_formato(total_victimas)}")
                input("\nPresione Enter para continuar y regresar al menú...")

            case 4:
              #M3 Regiones de México y su cultura de prevención (%) con Mapa Coroplético
              #Importamos la librería de geopandas,ya que amplía las funciones de la popular herramienta Pandas para permitir el trabajo con datos geoespaciales
              import geopandas as gpd
              #Modulo 2
              limpiar_pantalla()
              print("="*100)
              print("\n[OPCIÓN 4] REGIONES DE MÉXICO Y SU CULTURA DE PREVENCIÓN: \n")
              print("="*100)

              #M2 cálculos originales con factor de expansión
              responsables = df_ciberseguridad[df_ciberseguridad['realiza_medidas_seguridad'] == 1]
              pob_responsable = responsables.groupby(['nombre_estado'])['factor_expansion'].sum()
              pob_total = df_ciberseguridad.groupby(['nombre_estado'])['factor_expansion'].sum()
              porcentaje_por_estado = (pob_responsable / pob_total) * 100
              estados = porcentaje_por_estado.sort_values(ascending=False)

              #M3 Convertir mis resultados a un DataFrame para poder crear mi mapa Coroplético
              df_resultados = estados.reset_index()
              df_resultados.columns = ['nombre_estado', 'Porcentaje']

              #M3 cargar el mapa geográfico mediante la URL que encontre en internet
              url = "https://gist.githubusercontent.com/walkerke/76cb8cc5f949432f9555/raw/363c297ce82a4dcb9bdf003d82aa4f64bc695cf1/mx.geojson"
              gdf = gpd.read_file(url)

              #Convertir los nombres del mapa a MAYÚSCULAS para que se emparejen correctamente, ya que los de mi base de datos estan en mayúsculas
              gdf['name'] = gdf['name'].str.upper().str.strip()
              #Quitar acentos del mapa de internet (ej: QUERÉTARO -> QUERETARO), ya que en mi base de datos los estados no tienen acentos
              gdf['name'] = gdf['name'].str.normalize('NFKD').str.encode('ascii', errors='ignore').str.decode('utf-8')
              # Transformamos los nombres del mapa de internet para que calcen con el estándar del INEGI
              #Se tuvo que ocupar de un diccionario porque el mapa de internet tenia dos nombre que no coincidan con mi base de datos y me estaba dando valores nulos.
              diccionario_estados = {
                  'DISTRITO FEDERAL': 'CIUDAD DE MEXICO',
                  'MEXICO': 'MEXICO',
                  'COAHUILA': 'COAHUILA DE ZARAGOZA',
                  'MICHOACAN': 'MICHOACAN DE OCAMPO',
                  'VERACRUZ': 'VERACRUZ DE IGNACIO DE LA LLAVE'

              }
              #Reemplazamos los nombres de los estados de la base de datos del mapa de internet
              gdf['name'] = gdf['name'].replace(diccionario_estados)
              #Merge nos ayuda a unir el mapa dado de internet con mi columna de nombre_estado con la columna de nombre name (base de datos del mapa)
              gdf_unido = gdf.merge(df_resultados, left_on='name', right_on='nombre_estado', how='left')

              #Dibujar el mapa Coroplético avanzado con la librería Matplotlib
              fig, ax = plt.subplots(figsize=(14,11))

              gdf_unido.plot(
                  column='Porcentaje',
                  cmap='Greens',           # Degradado de color rojo (más oscuro = mayor adopción)
                  edgecolor='black',     # Bordes de los estados en negro
                  linewidth=0.5,
                  legend=True,           # Mostrar la barra de escala
                  legend_kwds={'label': "Porcentaje de Adopción (%)", 'orientation': "horizontal", 'pad': 0.02},
                  ax=ax
              )

              #Ciclo para agregar las etiquetas (nombre del estado y el pocrentaje)
              for idx, row in gdf_unido.iterrows():
                  #Se usa representative_point() para asegurar que la coordenada caiga dentro del estado
                  punto_interno = row['geometry'].representative_point()

                  #Formateamos el texto: Nombre del estado y porcentaje con 2 decimales
                  texto_etiqueta = f"{row['name']}\n{row['Porcentaje']:.2f}%"

                  #Dibujar el texto sobre el mapa
                  ax.annotate(
                      text=texto_etiqueta,
                      xy=(punto_interno.x, punto_interno.y),
                      horizontalalignment='center',
                      verticalalignment='center',
                      fontsize=4.5,
                      color='black',
                      weight='bold',
                      #Se agrega un sutil contorno blanco a las letras para que se lean sobre fondos oscuros
                      path_effects=[plt.matplotlib.patheffects.withStroke(linewidth=1.5, foreground="white")]
                  )

              plt.title('Mapa Coroplético: Cultura de Prevención por Entidad Federativa', fontsize=12, fontweight='bold')
              #ax.axis('off') oculta las coordenadas de latitud y longitud
              ax.axis('off')
              plt.tight_layout()
              plt.show()

              plt.pause(0.1)
              input("\nPresione Enter para continuar y regresar al menú...")

            case 5:
                #Cultura de seguridad por sexo (%)
                limpiar_pantalla()
                print("="*100)
                print("\n[OPCIÓN 5]CULTURA DE SEGURIDAD POR SEXO (%): \n")
                print("="*100)
                #M3 Columnas cruzadas
                #Crear una tabla cruzada(calculo de frecuencias de población real(usando la columna de factor de expansión) por sexo y medidas de seguridad)
                tabla_cruzada = pd.crosstab(
                    index=df_ciberseguridad['sexo'],
                    columns=df_ciberseguridad['realiza_medidas_seguridad'],
                    values=df_ciberseguridad['factor_expansion'],
                    aggfunc='sum'
                ).fillna(0)
                #Renombramos la columna de si realiza medidas de seguridad o no para que se pueda entender mejor.
                tabla_cruzada.columns = ['Sí realiza medidas', 'No realiza medidas','No responde / No sabe']
                #Convertimos los valores absolutos a porcentajes relativos por cada fila.
                tabla_porcentajes = tabla_cruzada.div(tabla_cruzada.sum(axis=1), axis=0) * 100
                #Esto mapea el índice para que en la gráfica diga Hombre y Mujer en lugar de 1 y 2 (nomenclatura de la base de datos del INEGI que ocupe).
                tabla_porcentajes.index = tabla_porcentajes.index.map({1: 'Hombre', 2: 'Mujer'})
                #Creamos la grafica
                #Creamos la grafica guardando los contenedores de las barras en ax.
                ax = tabla_porcentajes.plot(kind='bar', figsize=(12, 6), color=['#88E788','#FF9999', '#FFB347',])

                plt.title('Porcentaje de la Cultura de Prevención por Sexo', fontsize=12, fontweight='bold')
                plt.xlabel('Sexo / Género', color='#00008B', fontsize=11)
                plt.ylabel('Porcentaje (%)', color='#00008B', fontsize=11)
                plt.xticks(rotation=0)
                plt.legend(loc='upper right')

                #Usamos bar_label en vez de patches, descubri esta función cuando hice mi postwork 2 .
                #Esta funcion recibe como parámetro un "contenedor" de barras y se encarga de dibujar automáticamente las etiquetas de texto con sus valores correspondientes.
                for contenedor in ax.containers:
                    ax.bar_label(contenedor, fmt='%.2f%%', label_type='edge', fontweight='bold', fontsize=10)

                plt.tight_layout()
                plt.show()
                plt.pause(0.1)
                input("\nPresione Enter para continuar y regresar al menú...")

            case 6:
                #Efectividad
                limpiar_pantalla()
                print("="*100)
                print("\n[OPCIÓN 6] EFECTIVIDAD DE LAS MEDIDAS Y ANÁLISIS DE DISTRIBUCIÓN (BOXPLOT):\n")
                print("="*100)
                #M3 importamos la libreria seaborn para crear mis boxplot
                import seaborn as sns
                #Como tenemos 5 columnas de incidentes debemos almacenarlas en una lista para nuestro analisis
                columnas_delitos = ['incidente_hackeo_correo', 'incidente_hackeo_redes', 'incidente_ransomware', 'incidente_malware_conperdida', 'incidente_malware_general']
                sufrio_alguno = df_ciberseguridad[columnas_delitos].eq(1).any(axis=1)
                df_ciberseguridad['fue_hackeado_temp'] = sufrio_alguno.map({True: 1, False: 2})
                #Si la persona originalmente no contestó la sección de medidas (9), mantenemos el 9 en el análisis
                df_ciberseguridad.loc[df_ciberseguridad['realiza_medidas_seguridad'] == 9, 'fue_hackeado_temp'] = 9
                #EVALUACIÓN CON FACTOR DE EXPANSIÓN (Prueba A/B Conceptual) (los que si toman alguna medida de seguridad/ no toman alguna medida de seguridad)
                # Cruzamos medidas de seguridad con nuestra nueva columna de incidentes unificados
                tabla_efectividad = pd.crosstab(
                    index=df_ciberseguridad['realiza_medidas_seguridad'],
                    columns=df_ciberseguridad['fue_hackeado_temp'],
                    values=df_ciberseguridad['factor_expansion'],
                    aggfunc='sum'
                ).fillna(0)
                #Renombramos las filas y columnas en el orden estricto del INEGI (1=Sí, 2=No, 9=No responde)
                tabla_efectividad.index = ['Sí realiza medidas de seguridad', 'No realiza medidas de seguridad', 'No responde']
                tabla_efectividad.columns = ['Sí sufrió incidente', 'No sufrió incidente', 'No sabe']
                #Mostramos los porcentajes relativos por fila
                porcentajes_efectividad = tabla_efectividad.div(tabla_efectividad.sum(axis=1), axis=0) * 100
                print("\nTasa real de incidentes experimentados en la población según su prevención:")
                for condicion in porcentajes_efectividad.index:
                #.loc sirve para buscar, seleccionar y modificar bloques de datos en un DataFrame usando
                #las etiquetas de las filas y los nombres de las columnas, o mediante condiciones lógicas (booleanas).
                    tasa = porcentajes_efectividad.loc[condicion, 'Sí sufrió incidente']
                    print(f" - Usuarios [{condicion}]: {tasa:.2f}% de la población sufrió al menos un ciberdelito")

                #Chi2:valor estadístico Chi-cuadrada.
                #dof: Los grados de libertad (Degrees of Freedom).
                #esperados: La matriz de frecuencias teóricas esperadas.
                chi2, p_valor, dof, esperados = stats.chi2_contingency(tabla_efectividad)
                print(f"\nSignificancia estadística del experimento (Valor p): {p_valor:.6f}")
                if p_valor < 0.05:
                   print("Resultado estadístico: ALTAMENTE SIGNIFICATIVO.")
                    #Evaluamos cuál tasa es mayor para dar la conclusión real de mis datos
                   tasa_si = porcentajes_efectividad.loc['Sí realiza medidas de seguridad', 'Sí sufrió incidente']
                   tasa_no = porcentajes_efectividad.loc['No realiza medidas de seguridad', 'Sí sufrió incidente']

                   if tasa_si > tasa_no:
                      print("⚠️Conclusión: Se observa la paradoja de ciberseguridad. Los usuarios que toman medidas reportan mayor tasa de incidentes, asociado a una mayor capacidad de detección y exposición digital.")
                   else:
                       print("✅Conclusión: Las medidas de seguridad reducen eficazmente el riesgo real de sufrir ciberdelitos.")
                else:
                  print("⚠️Resultado estadístico: NO SIGNIFICATIVO. No hay diferencia probabilística clara entre los grupos.")

                #Grafico BOXPLOT
                #Analizamos cómo se distribuye la EDAD de las víctimas según implementen o no una medida de segurdiad (prevención)
                plt.figure(figsize=(8, 5))
                #Mapeamos los códigos numéricos a texto para que los ejes de la gráfica sean legibles
                df_temporal = df_ciberseguridad.copy()
                df_temporal['Prevención'] = df_temporal['realiza_medidas_seguridad'].map({1: 'Sí se protege', 2: 'No se protege', 9: 'No responde/ No sabe'})
                #Invocamos la función boxplot de Seaborn
                sns.boxplot(
                    x='Prevención',
                    y='edad',
                    data=df_temporal,
                    palette=['#88E788', '#FF9999', '#D3D3D3'],
                    linewidth=1.5
                )

                plt.title('Distribución de Edades según la Cultura de Prevención', fontsize=12, fontweight='bold')
                plt.xlabel('Condición de Seguridad del Usuario', color='#00008B', fontsize=11)
                plt.ylabel('Edad de los Usuarios (Años)', color='#00008B', fontsize=11)
                plt.tight_layout()
                plt.show()
                plt.pause(0.1)
                input("\nPresione Enter para continuar y regresar al menú...")

            case 7:
                #Percepción vs Realidad
                #axis=1 columnas y axis=0 filas
                limpiar_pantalla()
                print("="*100)
                print("\n[OPCIÓN 7]PERCEPCIÓN VS REALIDAD: \n")
                print("="*100)
                #M3 Machine Learning No Supervisado con el algoritmo K-Means
                #Preparación y limpieza de datos para mi algoritmo de machine learning
                #Nos aseguramos de que exista la columna unificada de incidentes
                if 'fue_hackeado_temp' not in df_ciberseguridad.columns:
                    columnas_delitos = ['incidente_hackeo_correo', 'incidente_hackeo_redes', 'incidente_ransomware', 'incidente_malware_conperdida', 'incidente_malware_general']
                    sufrio_alguno = df_ciberseguridad[columnas_delitos].eq(1).any(axis=1)
                    df_ciberseguridad['fue_hackeado_temp'] = sufrio_alguno.map({True: 1, False: 2})
                    df_ciberseguridad.loc[df_ciberseguridad['realiza_medidas_seguridad'] == 9, 'fue_hackeado_temp'] = 9
                #Filtramos para quitar valores 'No responde/No saben' (9) y dejar datos puramente analíticos para el algoritmo que vienen siendo (1(sí) y 2(no))
                df_ml = df_ciberseguridad[(df_ciberseguridad['realiza_medidas_seguridad'].isin([1, 2])) &
                                        (df_ciberseguridad['fue_hackeado_temp'].isin([1, 2]))].copy()
                #Seleccionamos las columnas numéricas que procesará el modelo
                columnas_entrenamiento = ['edad', 'realiza_medidas_seguridad', 'fue_hackeado_temp']
                #Limpiamos con dropna porque como quitamos los valores 9, pueden quedar valores nulos y no sería bueno para nuestro modelo
                df_ml_limpio = df_ml[columnas_entrenamiento].dropna()
                #Escalamos los datos (Regla matemática obligatoria para que los rangos de edad no aplasten al 1 y 0)
                escalador = StandardScaler()
                datos_escalados = escalador.fit_transform(df_ml_limpio)
                #ALGORITMO NO SUPERVISADO: K-MEANS (Segmentación de perfiles)
                print("Entrenando modelo de Machine Learning No Supervisado con K-Means...")
                modelo_kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
                df_ml_limpio['Perfil_Riesgo'] = modelo_kmeans.fit_predict(datos_escalados)
                print("\n✅Modelado K-Means completado con éxito.")
                print("Distribución de la población en los perfiles identificados por el Modelo:")
                conteo_clusters = df_ml_limpio['Perfil_Riesgo'].value_counts()
                for cluster_id, cantidad in conteo_clusters.items():
                    print(f" - Perfil Automatizado #{cluster_id}: {cantidad:,} registros clasificados")
                print("\n" + "="*100)
                print("📝PROTOCOLO DE EVALUACIÓN DEL MODELO:")
                print("="*100)
                #Validación Cruzada (Cross Validation - 5 pliegues) utilizando un clasificador de referencia
                X_eval = df_ml_limpio[['edad', 'realiza_medidas_seguridad']]
                y_eval = df_ml_limpio['fue_hackeado_temp']
                #70% entrenamiento y 30% prueba
                X_train, X_test, y_train, y_test = train_test_split(
                      X_eval,
                      y_eval,
                      test_size=0.30,
                      random_state=42
                )

                #Activamos el balanceo de pesos para romper el sesgo del modelo ya que solo esta tomando los NO y corregir el AUC
                clasificador_prueba = LogisticRegression(class_weight='balanced', max_iter=1000)

                # Entrenamos únicamente con el 70% de los datos
                clasificador_prueba.fit(X_train, y_train)

                # Predecimos utilizando el 30% de los datos nuevos (Test) que el modelo jamás ha visto
                y_predichos = clasificador_prueba.predict(X_test)

                #Matriz de Confusión evaluada sobre los datos de prueba
                matriz = confusion_matrix(y_test, y_predichos)

                #Curva ROC y cálculo de la métrica AUC sobre los datos de prueba (Test)
                #Evaluamos la probabilidad de pertenecer a la clase positiva (víctima = 1)
                y_probabilidades = clasificador_prueba.predict_proba(X_test)[:, 0]
                fpr, tpr, _ = roc_curve(y_test, y_probabilidades, pos_label=1)
                roc_auc = auc(fpr, tpr)

                #Validación Cruzada de control matemático
                clasificador_cv = LogisticRegression(class_weight='balanced', max_iter=1000)
                puntuaciones_cv = cross_val_score(clasificador_cv, X_eval, y_eval, cv=5, scoring='accuracy')
                print(f"➡️Exactitud media por Validación Cruzada (5-Folds): {puntuaciones_cv.mean()*100:.2f}%")
                print(f"➡️Área Bajo la Curva ROC (AUC) en Datos de Prueba (30%): {roc_auc:.4f}")
                print("="*100)

                # Creamos un lienzo dividido en 1 fila y 3 columnas
                #Porque realizare 3 gráficos
                fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(22, 6))

                #GRÁFICO 1: Mapa de Calor para la Matriz de Confusión
                sns.heatmap(
                            matriz, annot=True, fmt="d", cmap="Blues",
                            xticklabels=['Predijo Sí', 'Predijo No'], yticklabels=['Real Sí', 'Real No'], ax=ax1
                        )
                ax1.set_xlabel("Predicción", fontsize=11, fontweight='bold')
                ax1.set_ylabel("Valor Real", fontsize=11, fontweight='bold')
                ax1.set_title("Matriz de Confusión del Modelo", fontsize=12, fontweight='bold')

                #GRÁFICO 2: Segmentación Geométrica K-Means
                df_test_visual = df_ml_limpio.loc[X_test.index]
                scatter = ax2.scatter(
                    df_test_visual['edad'], df_test_visual['fue_hackeado_temp'],
                    c=df_test_visual['Perfil_Riesgo'], cmap='coolwarm', alpha=0.6, edgecolors='k', s=50
                )
                ax2.set_xlabel('Edad del Usuario (Años)', fontsize=11)
                ax2.set_ylabel('Condición de Incidente (1=Sí, 2=No)', fontsize=11)
                ax2.set_title('Segmentación No Supervisada (K-Means)', fontsize=12, fontweight='bold')
                cbar = fig.colorbar(scatter, ax=ax2)
                cbar.set_label('Perfiles Identificados por la IA', rotation=270, labelpad=15)

                #GRÁFICO 3:Curva ROC / AUC
                ax3.plot(fpr, tpr, color='darkorange', lw=2, label=f"AUC = {roc_auc:.2f}")
                ax3.plot([0, 1], [0, 1], color='navy', lw=2, linestyle="--") # Línea base diagonal de la aleatoriedad
                ax3.set_xlabel("Tasa de Falsos Positivos", fontsize=11)
                ax3.set_ylabel("Tasa de Verdaderos Positivos", fontsize=11)
                ax3.set_title("Curva ROC del Modelo", fontsize=12, fontweight='bold')
                ax3.legend(loc="lower right")

                plt.tight_layout()
                plt.show()
                plt.pause(0.1)

                input("\nPresione Enter para continuar y regresar al menú...")


            case 8:
                #Salir del programa
                print("="*100)
                print("\n[OPCIÓN 8]SALISTE DEL SISTEMA")
                print("\n¡Análisis finalizado con éxito!")
                print("="*100)
                ejecutando = False

            case _:
                print("\n[ALERTA] Opción fuera de rango. Elige un número del 1 al 8.")

    except ValueError:
        print("\n[ERROR DE ENTRADA] No escribiste un número válido. Por favor, digita un entero (1-8).")
        print(f"\n[ERROR REAL DETECTADO]: {e}")
        input("\nPresione Enter para continuar y regresar al menú...")
    except KeyError as e:
        print(f"\n[ERROR DE DATOS] Columna faltante en la base de datos: {e}")
    except Exception as e:
        print(f"\n[ERROR INESPERADO] Ocurrió una anomalía en el sistema: {e}")



```
